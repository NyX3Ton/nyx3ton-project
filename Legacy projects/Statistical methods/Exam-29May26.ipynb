```python
import numpy as np
from scipy import stats
import math

data = [6, 8, 2, 3, 6, 9, 1, 3, 4, 2, 9, 5, 1, 4, 7, 3, 8, 7, 7, 9]

alpha = 0.05

x = np.array(data, dtype=float)

n = len(x)
mean = np.mean(x)
s = np.std(x, ddof=1)   # vyberova smerodajna odchylka
df = n - 1

t_crit = stats.t.ppf(1 - alpha/2, df)
margin = t_crit * s / np.sqrt(n)

lower = mean - margin
upper = mean + margin

print(f"n = {n}")
print(f"Priemer = {mean:.3f}")
print(f"Smerodajna odchylka = {s:.3f}")
print(f"t kriticka hodnota = {t_crit:.6f}")
print(f"Dolna hranica = {lower:.3f}")
print(f"Horna hranica = {upper:.3f}")
```

```python
klientky = [214, 220, 211, 217, 209, 220, 216, 217, 219, 214, 217, 219, 217, 213, 215, 221, 219, 215, 216, 211, 218, 207]
klienti = [74, 111, 106, 114, 91, 99, 118, 102, 123, 96, 79, 94, 91, 103, 87, 99, 118, 125, 84, 110, 107, 110]

alpha = 0.05

n1 = len(klientky)
n2 = len(klienti)

mean1 = np.mean(klientky)
mean2 = np.mean(klienti)

var1 = np.var(klientky, ddof=1)
var2 = np.var(klienti, ddof=1)

# F-test, vacsi rozptyl / mensi rozptyl
F = max(var1, var2) / min(var1, var2)

# Welchov t-test
t_stat = (mean1 - mean2) / math.sqrt(var1/n1 + var2/n2)

# Welchove stupne volnosti
df = (var1/n1 + var2/n2)**2 / ((var1/n1)**2/(n1-1) + (var2/n2)**2/(n2-1))

p_value_t = 2 * stats.t.sf(abs(t_stat), df)

print(f"Priemer klientiek = {mean1:.3f}")
print(f"Priemer klientov = {mean2:.3f}")
print(f"Rozptyl klientiek = {var1:.3f}")
print(f"Rozptyl klientov = {var2:.3f}")
print(f"F statistika = {F:.3f}")
print(f"t statistika = {t_stat:.3f}")
print(f"df Welch = {df:.3f}")
print(f"p-hodnota t-testu = {p_value_t:.3e}")
```

```python
data = [6, 9, 7, 3, 8, 6, 4, 7, 7, 6, 4, 9, 2, 6, 6, 3, 4, 1, 2, 5, 7, 6]

alpha = 0.05
mu0 = 8

n = len(data)
mean = np.mean(data)
s = np.std(data, ddof=1)

t_stat = (mean - mu0) / (s / math.sqrt(n))
t_crit = stats.t.ppf(alpha, n - 1)
p_value = stats.t.cdf(t_stat, n - 1)

print(f"n = {n}")
print(f"Priemer = {mean:.3f}")
print(f"Smerodajna odchylka = {s:.3f}")
print(f"Testovacia statistika = {t_stat:.3f}")
print(f"Kriticky kvantil = {t_crit:.3f}")
print(f"p-hodnota = {p_value:.6f}")

if t_stat < t_crit:
    print("H0 zamietame.")
else:
    print("H0 nezamietame.")
```
